```python
def histogram_manual(img, L=256):
    h = np.zeros(L, dtype=np.int64)
    for y in range(img.shape[0]):
        for x in range(img.shape[1]):
            h[img[y, x]] += 1
    return h
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

NIM = '244107020024'
N = int(NIM[-3:])
PARAM = {
    'bins': 2 ** ((N % 4) + 3),
    'clip': round(1.0 + (N % 5) * 0.5, 1),
    'tile': (N % 4) + 2,
    'L': 2 ** ((N % 3) + 1),
}

gray = cv2.imread(CONTOH + '/lena_lc.jpg', cv2.IMREAD_GRAYSCALE)

h_manual = histogram_manual(gray)
h_numpy, _ = np.histogram(gray, bins=256, range=(0, 256))
h_cv = cv2.calcHist([gray], [0], None, [256], [0, 256]).flatten()

print('selisih manual vs numpy :', np.max(np.abs(h_manual - h_numpy)))
print('selisih manual vs cv    :', np.max(np.abs(h_manual - h_cv)))

hasil_cv = cv2.equalizeHist(gray)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title(NIM + ' - Citra Asli')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(hasil_cv, cmap='gray')
plt.title(NIM + ' - Equalization')
plt.axis('off')

plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

berkas = BASE + '/KTM1d.jpeg'
bgr = cv.imread(berkas, cv.IMREAD_COLOR)

if bgr is None:
  raise FileNotFoundError(f'File gak ketemu: {berkas}.')
print('Shape gambar:', bgr.shape)

kanal = list(cv.split(bgr))
for i in range(3):
  kanal[i] = cv.equalizeHist(kanal[i])
he_rgb = cv.merge(kanal)

ycrcb = cv.cvtColor(bgr, cv.COLOR_BGR2YCrCb)
y, cr, cb = cv.split(ycrcb)
he_y = cv.cvtColor(
    cv.merge([cv.equalizeHist(y), cr, cb]), cv.COLOR_YCrCb2BGR
)

hsv = cv.cvtColor(bgr, cv.COLOR_BGR2HSV)
h, sat, v = cv.split(hsv)
he_v = cv.cvtColor(cv.merge([h, sat, cv.equalizeHist(v)]), cv.COLOR_HSV2BGR)

lab = cv.cvtColor(bgr, cv.COLOR_BGR2LAB)
l, a, b = cv.split(lab)
he_l = cv.cvtColor(cv.merge([cv.equalizeHist(l), a, b]), cv.COLOR_LAB2BGR)

judul = ['Asli', 'Ekualisasi B, G, R', 'Kanal Y', 'Kanal V', 'Kanal L']
citra = [bgr, he_rgb, he_y, he_v, he_l]

plt.figure(figsize=(18, 4))
for i, (c, t) in enumerate(zip(citra, judul), 1):
  plt.subplot(1, 5, i)
  plt.imshow(cv.cvtColor(c, cv.COLOR_BGR2RGB))
  plt.title(NIM + ' - ' + t, fontsize=9)
  plt.axis('off')
plt.show()


def geser_hue(asli, hasil):
  h1 = cv.cvtColor(asli, cv.COLOR_BGR2HSV)[:, :, 0].astype(np.int16)
  h2 = cv.cvtColor(hasil, cv.COLOR_BGR2HSV)[:, :, 0].astype(np.int16)
  d = np.abs(h1 - h2)
  d = np.minimum(d, 180 - d)
  return d.mean()


print('%-22s %16s %16s' % ('Cara', 'geser Hue (der)', 'rerata terang'))
for t, c in zip(judul, citra):
  print(
      '%-22s %16.2f %16.1f'
      % (
          t,
          geser_hue(bgr, c) * 2,
          cv.cvtColor(c, cv.COLOR_BGR2GRAY).mean(),
      )
  )
```

1. Cara dengan Pergeseran Hue Paling Besar:
utput yang dicetak (print) di bagian bawah kodenya. Metode yang biasanya menghasilkan perubahan angka pergeseran Hue (derajat) terbesar adalah Ekualisasi kanal B, G, R karena memodifikasi informasi warna pada setiap kanal secara terpisah tanpa mempertahankan konsistensi warnanya.
2. Penyebab Pergeseran Hue Kanal Terang Tidak Persis Nol:
Walaupun kanal warna asli (seperti Cr, Cb, Saturation, atau komponen A dan B) tidak terpengaruh saat ekualisasi kanal terang, proses konversi kembali ke ruang warna BGR melibatkan perhitungan matematis, pembulatan nilai desimal, dan pemotongan terhadap nilai piksel agar tetap dalam rentang 0 hingga 255. Ini mengakibatkan sedikit distorsi pada nilai Hue.  
3. Kanal yang Paling Sesuai untuk Foto KTM:
Kanal Y (YCrCb) atau Kanal L (LAB) umumnya lebih cocok karena dapat meningkatkan kontras teks dan kejelasan informasi pada kartu identitas tanpa merusak keseimbangan warna aslinya secara berlebihan.
4. Perbandingan dengan CLAHE:
Dengan menerapkan CLAHE menggunakan parameter clipLimit = PARAM['clip'] dan tileGridSize = (PARAM['tile'], PARAM['tile']), perubahan pada Hue dan rata-rata kecerahan akan sedikit bervariasi dibandingkan dengan ekualisasi global, di mana CLAHE menghindari over-amplification di area yang sangat terang, sehingga detail teks pada KTM sering kali lebih jelas terbaca.

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

berkas = BASE + '/KTM1a.jpeg'
gray = cv.imread(berkas, cv.IMREAD_GRAYSCALE)

if gray is None:
  raise FileNotFoundError(f'File tidak ditemukan di path: {berkas}.')

h = np.histogram(gray, bins=256, range=(0, 256))[0]
p = h / gray.size
cdf = np.cumsum(p)
manual_eq = np.round(cdf * 255).astype(np.uint8)
img_manual = manual_eq[gray]

img_cv = cv.equalizeHist(gray)

psnr_val = cv.PSNR(gray, img_cv)
print('PSNR:', psnr_val)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title(NIM + ' - Asli', fontsize=9)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_cv, cmap='gray')
plt.title(NIM + ' - EqualizeHist', fontsize=9)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.plot(np.histogram(gray, bins=256, range=(0, 256))[0], color='black', label='Asli')
plt.plot(np.histogram(img_cv, bins=256, range=(0, 256))[0], color='red', label='Equalize')
plt.title(NIM + ' - Histogram', fontsize=9)
plt.legend()

plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

berkas = BASE + '/KTM1a.jpeg'
gray = cv.imread(berkas, cv.IMREAD_GRAYSCALE)

if gray is None:
  raise FileNotFoundError(f'File tidak ditemukan di path: {berkas}.')

img_cv = cv.equalizeHist(gray)

clahe = cv.createCLAHE(
    clipLimit=PARAM['clip'], tileGridSize=(PARAM['tile'], PARAM['tile'])
)
img_clahe = clahe.apply(gray)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_cv, cmap='gray')
plt.title(NIM + ' - Ekualisasi Global', fontsize=9)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_clahe, cmap='gray')
plt.title(NIM + ' - CLAHE (clip: ' + str(PARAM['clip']) + ')', fontsize=9)
plt.axis('off')
plt.show()

base_clip = PARAM['clip']
clip_vals = [
    max(0.5, base_clip - 1.0),
    base_clip,
    base_clip + 1.0,
    base_clip + 2.0,
]

plt.figure(figsize=(16, 4))
for idx, c_val in enumerate(clip_vals, 1):
  custom_clahe = cv.createCLAHE(
      clipLimit=c_val, tileGridSize=(PARAM['tile'], PARAM['tile'])
  )
  res_img = custom_clahe.apply(gray)

  plt.subplot(1, 4, idx)
  plt.imshow(res_img, cmap='gray')
  plt.title(f'{NIM} - Clip: {c_val:.1f}', fontsize=9)
  plt.axis('off')
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np


def floyd_steinberg(gray, L=2):
  img = gray.astype(np.float32).copy()
  step = 255.0 / (L - 1)
  H, W = img.shape
  for y in range(H):
    for x in range(W):
      lama = img[y, x]
      baru = np.round(lama / step) * step
      img[y, x] = baru
      error = lama - baru
      if x + 1 < W:
        img[y, x + 1] += error * 7 / 16
      if x > 0 and y + 1 < H:
        img[y + 1, x - 1] += error * 3 / 16
      if y + 1 < H:
        img[y + 1, x] += error * 5 / 16
      if x + 1 < W and y + 1 < H:
        img[y + 1, x + 1] += error * 1 / 16
  return np.clip(img, 0, 255).astype(np.uint8)


berkas = BASE + '/KTM1b.jpeg'
bgr = cv.imread(berkas)

if bgr is None:
  raise FileNotFoundError(f'File tidak ditemukan di path: {berkas}.')

bgr_dither = bgr.copy()
for i in range(3):
  bgr_dither[:, :, i] = floyd_steinberg(bgr[:, :, i], L=PARAM['L'])

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(bgr, cv.COLOR_BGR2RGB))
plt.title(NIM + ' - KTM1b.jpg Asli', fontsize=9)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(bgr_dither, cv.COLOR_BGR2RGB))
plt.title(NIM + f' - Dithering L={PARAM["L"]}', fontsize=9)
plt.axis('off')
plt.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np


def floyd_steinberg(gray, L=2):
  img = gray.astype(np.float32).copy()
  step = 255.0 / (L - 1)
  H, W = img.shape
  for y in range(H):
    for x in range(W):
      lama = img[y, x]
      baru = np.round(lama / step) * step
      img[y, x] = baru
      error = lama - baru
      if x + 1 < W:
        img[y, x + 1] += error * 7 / 16
      if x > 0 and y + 1 < H:
        img[y + 1, x - 1] += error * 3 / 16
      if y + 1 < H:
        img[y + 1, x] += error * 5 / 16
      if x + 1 < W and y + 1 < H:
        img[y + 1, x + 1] += error * 1 / 16
  return np.clip(img, 0, 255).astype(np.uint8)


berkas = BASE + '/KTM1a.jpeg'
gray = cv.imread(berkas, cv.IMREAD_GRAYSCALE)

if gray is None:
  raise FileNotFoundError(f'File tidak ditemukan di path: {berkas}.')

dither_langsung = floyd_steinberg(gray, L=2)

eq = cv.equalizeHist(gray)
eq_dither = floyd_steinberg(eq, L=2)

plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title(NIM + ' - Asli Grayscale', fontsize=8)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(dither_langsung, cmap='gray')
plt.title(NIM + ' - Dither Langsung', fontsize=8)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(eq_dither, cmap='gray')
plt.title(NIM + ' - Ekualisasi + Dither', fontsize=8)
plt.axis('off')

plt.show()

'''
- Urutan mana yang menghasilkan teks pada KTM lebih mudah dibaca?
Urutan yang membuat teks pada KTM lebih mudah dibaca adalah melakukan ekualisasi histogram terlebih dahulu, sebelum melanjutkan dengan dithering
Citra `KTM1a.jpg` diambil dalam kondisi gelap sehingga kontrasnya sangat rendah[cite: 4]. Jika dithering diterapkan segera, detail yang gelap akan hilang dan teks di kartu akan tampak kabur. Dengan melakukan ekualisasi histogram terlebih dahulu, distribusi intensitas piksel diperluas dan kontrasnya meningkat, sehingga saat proses dithering mengurangi level warna, bentuk teks dan tepi pada kartu tetap dapat dipertahankan dengan jelas
'''
```
