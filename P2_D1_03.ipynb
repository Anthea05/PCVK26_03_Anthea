```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

# Membuat gambar background hitam ukuran 512 x 512
black_img2 = np.zeros((512, 512, 3), dtype=np.uint8)

pts = np.array([
    [250, 250],
    [450, 350],
    [250, 450],
    [125, 350]
], np.int32)

pts = pts.reshape((-1, 1, 2))

cv.polylines(
    black_img2,
    [pts],
    isClosed=True,
    color=(0, 255, 0),
    thickness=5
)

plt.imshow(black_img2)
plt.show()
```
